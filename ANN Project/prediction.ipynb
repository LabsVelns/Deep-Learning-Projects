```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from keras.models import  load_model
import streamlit as st
import pickle
import datetime
```

```python
model = load_model('ann_model.h5')

## Load the pickle file
with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)

with open('label_encoder_gender.pkl', 'rb') as f:
    label_encoder = pickle.load(f)  

with open('OneHotGeo.pkl', 'rb') as f:
    onehot_encoder = pickle.load(f) 
```

```python
# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
geo_encoded = onehot_encoder.transform([[input_data['Geography']]]).toarray()
geo_df = pd.DataFrame(geo_encoded, columns=onehot_encoder.get_feature_names_out(['Geography']))
geo_df
```

```python
df = pd.DataFrame([input_data])
df = pd.concat([geo_df, df.drop(['Geography'], axis=1)], axis=1)
df
```

```python
## Gender Label Encoding
df['Gender'] = label_encoder.transform(df['Gender'])
df
```

```python
scaled_data = scaler.transform(df)
scaled_data
```

```python
## model Prediction
prediction = model.predict(scaled_data)
```
